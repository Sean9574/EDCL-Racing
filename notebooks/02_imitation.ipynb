# 2. Imitation-learning controllers
Baseline with each run seed, the three Keras Tuner searches, and retraining of each selected architecture with each run seed (design specification §8).

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))
```

```python
from racing import config, imitation

data = imitation.load_dataset()
for seed in config.RUN_SEEDS:
    print(imitation.train_baseline(data, seed=seed))
```

```python
for approach in config.TUNER_APPROACHES:
    summary = imitation.run_tuner(approach, data)
    print({k: v for k, v in summary.items() if k != 'chosen_hyperparameters'})
    for seed in config.RUN_SEEDS:
        print(imitation.retrain(approach, data, seed=seed))
```
