# 3. DAgger and speed push
Each imitation controller is refined by DAgger rounds and then pushed for speed by an evolution strategy grown from it, once per run seed (spec §8.4–8.5).

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))
```

```python
from racing import config, dagger, imitation


BRIEF_KEYS = ('approach', 'seed', 'level_reached', 'goal_reached', 'outcome', 'target_reduction',
              'generations', 'rounds', 'val_successes', 'val_accuracy', 'wall_time')


def brief(summary):
    """One line per run; the live progress and milestones are shown by scripts/run_notebooks.py."""
    for value in summary.values():
        if isinstance(value, dict) and 'approach' in value:
            brief(value)                      # a stage of a combined run, e.g. DAgger and speed push
    print({key: summary[key] for key in BRIEF_KEYS if key in summary}, flush=True)


data = imitation.load_dataset()
for approach in config.IMITATION_APPROACHES:
    for seed in config.RUN_SEEDS:
        brief(dagger.run(approach, seed, data, n_jobs=config.N_JOBS))
```
