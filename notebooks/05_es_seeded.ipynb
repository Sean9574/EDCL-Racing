# 5. Seeded evolution strategy
The ES started from the baseline trained with the same run seed (design specification §9.4).

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))
```

```python
from racing import config, es


def progress(report):
    if report['generation'] % 10 == 0 or report['leveled_up']:
        print(f"gen {report['generation']:4d}  level {report['level']}  best {report['best']:.3f}  "
              f"new finish {report['new_finish']:.2f}  at pace {report['whole_pace']:.2f}  "
              f"step {report['step_size']:.4f}")


for seed in config.RUN_SEEDS:
    print(f'--- seed {seed}')
    print(es.run_es(seed=seed, initial=es.trained_baseline(seed), approach='es_seeded',
                    on_generation=progress, n_jobs=config.N_JOBS))
```
